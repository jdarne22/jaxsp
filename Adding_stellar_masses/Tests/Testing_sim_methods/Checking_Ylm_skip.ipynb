```python
import numpy as np
import jax.numpy as jnp
from scipy.special import sph_harm_y
import sys
sys.path.append('/home/joshua/PhD_year_1/jaxsp/Adding_stellar_masses')
import Stellar_sim_funcs as SSF

L_max_out = 47   # whatever your sim uses
rng = np.random.default_rng(0)
theta = rng.uniform(0.1, np.pi - 0.1, size=10)   # off-pole
phi   = rng.uniform(0, 2*np.pi, size=10)

# scipy reference
lm_pairs = np.array([(L, M) for L in range(L_max_out) for M in range(-L, L+1)])
Y_sp, dY_sp = sph_harm_y(lm_pairs[:, 0, None], lm_pairs[:, 1, None],
                         theta[None, :], phi[None, :], diff_n=1)
dY_sp = dY_sp[:, :, 0]

# jax version
Y_jx, dY_jx = SSF.Ylm_dY_jax(L_max_out, jnp.asarray(theta), jnp.asarray(phi))

print("Y max abs diff:",  np.max(np.abs(np.asarray(Y_jx)  - Y_sp)))
print("dY max abs diff:", np.max(np.abs(np.asarray(dY_jx) - dY_sp)))
```
